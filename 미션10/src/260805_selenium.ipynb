### 날짜 버튼 클릭

```python
from selenium.webdriver.support.ui import Select
```

```python
# 크롬 브라우저 실행
driver = webdriver.Chrome()

# 페이지 열기
driver.get("https://workey.codeit.kr/ratings/index")

time.sleep(2)

# 연도별 구분자 확인
year_select_element = driver.find_element(By.NAME, "year")
year_select = Select(year_select_element)
# 연도를 2013년도로 변경
year_select.select_by_visible_text("2013")

# 월 별 구분자 확인
month_select_element = driver.find_element(By.NAME, "month")
month_select = Select(month_select_element)
# 월을 5월로 변경
month_select.select_by_visible_text("5")

# 참조 - search 버튼을 누를 때 쓰는 코드
search_button = driver.find_element(By.CLASS_NAME, "search-btn")
search_button.click()
```

### 명시적 기다림

```python
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
```

```python
driver = webdriver.Chrome()
driver.get("https://workey.codeit.kr/ratings/index")

wait = WebDriverWait(driver, 10)

year_select_element = wait.until(
    EC.presence_of_element_located((By.NAME, "year"))
)

# 연도별 구분자 확인
year_select_element = driver.find_element(By.NAME, "year")
year_select = Select(year_select_element)
# 연도를 2013년도로 변경
year_select.select_by_visible_text("2013")
```
