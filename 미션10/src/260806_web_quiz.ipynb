### Q7. 티비랭킹닷컴에 있는 모든 데이터를 가져와 '시청률.xlsx' 파일로 저장하세요. 데이터는 2010년부터 2018년까지 데이터를 모두 포함하며, 헤더 이름은 기간, 순위, 채널, 프로그램, 시청률로 정하고, 기간은 YYYY년 MM월 W주차 형태로 출력하세요.

```python
%pip install openpyxl
```

```python
import pandas as pd
from openpyxl import Workbook


wb = Workbook()


ws = wb.active
ws.title = "시청률"

ws.append(["기간", "순위", "채널", "프로그램", "시청률"])

session = requests.Session()

for year in range(2010, 2019):
    for month in range(1, 13):
        for week in range(5):
            
            url = f"https://workey.codeit.kr/ratings/index?year={year}&month={month}&weekIndex={week}"
            response = session.get(url)
            soup = BeautifulSoup(response.text, "html.parser")
            
            rows = soup.select("tr")[1:]
            
            for row in rows:
                columns = row.select("td")
                
                if len(columns) == 4:
                    ws.append([
                        f"{year}년 {month}월 {week+1}주차",
                        columns[0].get_text(strip=True),
                        columns[1].get_text(strip=True),
                        columns[2].get_text(strip=True),
                        columns[3].get_text(strip=True),
                    ])

wb.save("시청률.xlsx")
```

```python
wb = Workbook(write_only=True)

for year in range(2010, 2019):
    ws = wb.create_sheet(f'{year}년')
    ws.append(['기간', '순위', '채널', '프로그램', '시청률'])

    for month in range(1, 13):
        for weekIndex in range(0, 5):
            response = requests.get(f"https://workey.codeit.kr/ratings/index?year={year}&month={month}&weekIndex={weekIndex}")
            rating_page = response.text
            soup = BeautifulSoup(rating_page, 'html.parser')

            tr_tags = soup.select('tr')[1:]

            for tr_tag in tr_tags:
                td_tags = tr_tag.select('td')
                period = f"{year}년 {month}월 {weekIndex+1}주차"
                row = [
                    period,
                    td_tags[0].get_text(),
                    td_tags[1].get_text(),
                    td_tags[2].get_text(),
                    td_tags[3].get_text()
                ]
                ws.append(row)

wb.save('시청률.xlsx')
```

### Q5. 오렌지 보틀 (https://workey.codeit.kr/orangebottle/index) 의 모든 지점의 이름, 주소, 전화번호를 '오렌지_보틀.xlsx'에 저장하세요. 헤더 행은 '지점 이름', '주소', '전화번호'로 저장하세요.

```python
url =  "https://workey.codeit.kr/orangebottle/index"
response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

wb = Workbook(write_only=True)
ws = wb.create_sheet()
ws.append(["지점 이름", "주소", "전화번호"])

branches = soup.select("div.branch")

for branch in branches:
    branch_name = branch.select_one("p.city").get_text(strip=True)
    address = branch.select_one("p.address").get_text(strip=True)
    phone = branch.select_one("span.phoneNum").get_text(strip=True)
    ws.append([branch_name, address, phone])

wb.save("오렌지 보틀.xlsx")
```
