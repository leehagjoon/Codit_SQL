```python
import requests
```

```python
url = "https://ko.wikipedia.org/wiki/%EC%9B%B9_%EC%8A%A4%ED%81%AC%EB%9E%98%ED%95%91"
```

```python
# 서버에 요청 보내기
response = requests.get(url)
```

```python
# 응답 확인
print("상태 코드 : ", response)
```

```python
# 거절이 된 이유를 출력
print("거절 이유 : ", response.text[:500])
```

```python
# 유저 에이전트 밝히기

header = {
    "User-Agent": "Mozilla/5.0 (compatible; EducationalScraper/1.0; +https://example.com)"
}

#서버에 요청 보내기
response = requests.get(url, headers=header)

print("상태 코드 : ", response.status_code)
```

```python
print(response.text)
```

```python
# 참조 - 파일 저장용 코드
with open("wikipedia_scraping.txt","w", encoding="utf-8") as f:
    f.write(response.text)
```

```python
# 변수 지정

html = response.text
```

```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(html, "html.parser")
```

```python
print(soup.title.get_text())
```

```python
# 제목 텍스트 가져오기 - select_one() 함수 사용
h1 = soup.select_one("h1#firstHeading")
print(h1.get_text())
```

```python
# 제목 텍스트 가져오기 - find() 함수 사용
title = soup.find("h1", id="firstHeading")
print(title.get_text())
```

```python
# 첫 단락 가져오기
first_p = soup.select_one("#mw-content-text p")
print(first_p.get_text(" ", strip=True)[:100])
```

```python
# 세번째 단락 가져오기
third_p = soup.select("#mw-content-text p")[2]
print(third_p.get_text(" ", strip=True))
```

### 링크 가져오기

```python
# 본문 영역 안에 있는 모든 a 태그 선택
items = soup.select("#mw-content-text a")
items
```

```python
# 하단 같이 보기 링크 깔끔하게 가져오기

h2 = soup.select_one("h2#같이_보기")
heading_div = h2.find_parent("div", class_="mw-heading")
ul = heading_div.find_next_sibling("ul")

items_2 = ul.select("li a")

for a in items_2:
    print(a.get_text(strip=True), '-', a.get("href"))
```
